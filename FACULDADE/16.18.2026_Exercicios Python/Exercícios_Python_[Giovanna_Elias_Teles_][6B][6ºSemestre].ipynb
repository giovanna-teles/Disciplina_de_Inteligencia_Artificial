## 2. Soma de Dois Números

Crie um programa que peça ao usuário dois números e exiba a soma
deles.

**Conceitos que caem aqui:** `input()`, `float()`, `variáveis`, `soma (+)` e `print()`

**Minha explicação:** Nesse código, primeiro pedimos dois números para o usuário utilizando o `input()`. O problema é que o `input()` sempre entende o que a pessoa digitou como texto `(string)`, mesmo que ela digite um número. Por isso utilizamos o `float()` para transformar esse texto em número e também permitir números quebrados, exemplo 2.5. Depois guardamos cada número em uma variável (como se fosse uma caixinha), n1 e n2, fazemos a soma utilizando o `+` e guardamos o resultado na variável soma. Por último, utilizamos o `print()` para mostrar o resultado para o usuário.

**Sugestão:** Caso você deseje um "mini sistema", ou seja, seguir o fluxo: usuário digita os valores -> sistema soma -> sistema mostra o resultado para o usuário -> sistema pergunta "Deseja um novo calculo de soma?" -> Se sim, volta pedindo para o usuário digitar um novo número, se não, sistema encerra. Teremos que utilizar os conceitos de `if`,`else`,`elif`,`While True`,`==`  e `break`. Eu acho mais interessante trabalharmos dessa forma, vou usar a oportunidade para praticar nesse exercício.Vou inserir uma trava também, caso o usuário digite um valor que não seja 0 ou 1, pedindo para ele digitar o número novamente.


```python
print("Bem vindo a mini calculadora de somar! :)")
print(" ")  # Isso aqui é apenas para pular linha

sim = 1
nao = 0

while True:  # Mantém a calculadora funcionando até o usuário decidir parar

    n1 = float(input("Digite o primeiro número: "))
    n2 = float(input("Digite o segundo número: "))

    soma = n1 + n2
    print("A soma é:", soma)

    print(" ")

    # Esse while serve somente para validar a escolha do usuário
    while True:
        print("Deseja um novo calculo de soma?")
        n3 = int(input("Digite 1 para SIM ou digite 0 para NÃO: "))

        if n3 == sim:
            print(" ")
            print("Ebaa!! Vamos fazer outro cálculo!")
            print(" ")
            break  # Sai desse while e volta para o cálculo

        elif n3 == nao:
            print(" ")
            print("Obrigado por usar a mini calculadora!")
            break  # Sai desse while e encerra o programa

        else: #Trava que mencionei,caso ele digite um valor que não é 0 ou 1
            print(" ")
            print("Opção inválida. Digite 1 para SIM ou 0 para NÃO.")
            print(" ")
            # Não tem break aqui!!!!!
            # Então ele pergunta novamente 1 ou 0.

    # Se o usuário escolheu 0, encerra a mini calculadora
    if n3 == nao:
        break
```

## 3. Cálculo da Área do Círculo

Solicite o raio de um círculo e calcule sua área. Use a fórmula área =
π * r².

**Conceitos que caem aqui:** `import`, `math.pi`, `float()`, ** e `round()`

**Minha explicação:** Nesse código aqui, vamos calcular a área de um círculo. Primeiro usamos o `import` math, que serve para importar a biblioteca math do Python, isso facilita bastante, pois essa biblioteca possui várias funções e valores matemáticos que podemos utilizar. Nesse caso, usamos o `math.pi`, que traz o valor do `π`. Depois pedimos o raio para o usuário e usamos o `**2` para elevar o raio ao quadrado. Por fim, fazemos a conta da área e usamos o `round()` para mostrar o resultado com apenas duas casas decimais. Se não, vai trazer o mundo inteiro dos números pós virgula desse calculo.

**Sugestão:** Vou utilizar os conceitos de `if`,`else`,`elif`,`While True`,`==`  e `break`, igual fiz no exercício 2 :). Para termos um mini sistema.



```python
import math
print("Bem vindo a mini calculadora de área do círculo! :)")
print(" ")

sim = 1
nao = 0

while True:
    print("Bem vindo a mini calculadora de área do círculo! :)")
    print(" ")

    raio = float(input("Digite o raio do círculo: "))

    area = math.pi * (raio ** 2)
    print("A área do círculo é:", round(area, 2))

    print(" ")

    # Enquanto o usuário não digitar 1 ou 0, continua perguntando
    while True:
        print("Deseja calcular a área de outro círculo?")
        n2 = int(input("Digite 1 para SIM ou digite 0 para NÃO: "))

        if n2 == sim:
            print(" ")
            print("Vamos calcular outro círculo!")
            print(" ")
            break

        elif n2 == nao:
            print(" ")
            print("Obrigada por usar a calculadora!")
            break

        else:
            print(" ")
            print("Opção inválida. Digite 1 para SIM ou 0 para NÃO.")
            print(" ")

    # Se o usuário escolheu 0, encerra a mini calculadora
    if n2 == nao:
        break
```

##4. Conversão de Temperatura
Converta uma temperatura fornecida em Celsius para Fahrenheit
usando a fórmula F = C * 9/5 + 32.

**Conceitos que caem aqui:** `input()`, `float()`, `divisão /`, `multiplicação *` e `print()`.

**Minha explicação:** Nesse código aqui, vamos converter uma temperatura de `Celsius` para `Fahrenheit`. Primeiro usamos o `input()` para pedir a temperatura e o `float()` para transformar o valor digitado em número. Depois usamos a fórmula `F = (C * 9 / 5) + 32` e guardamos o resultado na variável fahrenheit. É importante usar `/` na divisão e não `//`, porque o `/` mantém o resultado quebrado, enquanto o `//` faz uma divisão inteiraa e poderia deixar o cálculo errado. ## 8. Verificação de Palíndromo

Crie um programa que verifique se uma palavra é um palíndromo
(ex: "radar").

**Conceitos que caem aqui:**  `[::-1]`, `.lower()`, `.strip()`, `if`

**Minha explicação:** Palíndromo é uma palavra que fica igual quando lida de trás pra frente. Nesse código, primeiro pedimos uma palavra para o usuário. Usamos o `.lower()` para transformar as letras em minúsculas, assim se a pessoa digitar "Radar", fica "radar". Já o `.strip()` serve para tirar espaços que a pessoa possa colocar antes ou depois da palavra , resumindo usamos o `.lower()` e `.strip()` para não dar bug no nosso mini sistema . Depois usamos o `[::-1]` para inverter a palavra e o `if` compara as duas. Se forem iguais, é um palíndromo.

**Sugestão:** Vou utilizar os conceitos de `if`,`else`,`elif`,`While True`,`==`  e `break`, igual fiz no exercício 2 :). Para termos um mini sistema.



```python
sim = 1
nao = 0

while True:
    print("Programa para verificar se uma palavra é um palíndromo! :)")
    print(" ")

    palavra = input("Digite uma palavra: ")

    palavra_limpa = palavra.lower().strip()
    invertida = palavra_limpa[::-1]

    if palavra_limpa == invertida:
        print("É um palíndromo!")
    else:
        print("Não é um palíndromo")

    print(" ")

    while True:
        print("Deseja verificar outra palavra?")
        n2 = int(input("Digite 1 para SIM ou digite 0 para NÃO: "))

        if n2 == sim:
            print(" ")
            print("Vamos verificar outra palavra!")
            print(" ")
            break

        elif n2 == nao:
            print(" ")
            print("Obrigada por usar o programa!")
            break

        else:
            print(" ")
            print("Opção inválida. Digite 1 para SIM ou 0 para NÃO.")
            print(" ")

    if n2 == nao:
        break
```

## 9. Fatorial de um Número

Calcule o fatorial de um número fornecido pelo usuário.

**Conceitos que caem aqui:** `for` acumulando resultado.

**Minha explicação:**Fatorial é quando a gente pega um número e vai multiplicando ele pelos números que vêm antes até chegar no 1. Primeiro pedimos um número para o usuário. Depois usamos o for para ir passando pelos números anteriores até chegar no 1. A cada volta, fazemos uma multiplicação e guardamos o resultado em uma variável. Essa variável começa com 1, porque se começasse com 0, qualquer número multiplicado por ela daria 0. No final, temos o resultado do fatorial.

```python
sim = 1
nao = 0

while True:
    print("Programa para calcular o fatorial! :)")
    print(" ")

    numero = int(input("Digite um número para calcular o fatorial: "))

    fatorial = 1

    for i in range(1, numero + 1):
        fatorial = fatorial * i

    print("O fatorial de", numero, "é", fatorial)

    print(" ")

    while True:
        print("Deseja calcular outro fatorial?")
        n2 = int(input("Digite 1 para SIM ou digite 0 para NÃO: "))

        if n2 == sim:
            print(" ")
            print("Vamos calcular outro fatorial!")
            print(" ")
            break

        elif n2 == nao:
            print(" ")
            print("Obrigada por usar o programa!")
            break

        else:
            print(" ")
            print("Opção inválida. Digite 1 para SIM ou 0 para NÃO.")
            print(" ")

    if n2 == nao:
        break
```

12. Contagem de Caracteres

Conte o número de ocorrências de cada caractere em uma string
fornecida.

Conceitos que caem aqui: dicionário

Minha explicação: Primeiro temos um texto que queremos analisar (usuario digita). Depois o código vai passando por cada caractere, um por um. Usamos um dicionário para guardar o caractere e a quantidade de vezes que ele aparece. Se o caractere aparecer pela primeira vez, começamos a contagem em 1. Se ele aparecer novamente, aumentamos essa quantidade em 1. No final, conseguimos saber quantas vezes cada caractere apareceu

```python
texto = input("Digite uma palavra ou frase: ")

contagem = {}
for caractere in texto:
    if caractere in contagem:
        contagem[caractere] += 1
    else:
        contagem[caractere] = 1

for caractere, quantidade in contagem.items():
    print(caractere, "->", quantidade)
```

18. Número Primo

Crie um programa que verifique se um número fornecido é primo

Conceitos que caem aqui: %, for, break e if/else

Minha explicação: Primeiro pedimos um número para o usuário. Um número primo é aquele que só pode ser dividido por 1 e por ele mesmo. Depois usamos o for para testar se ele pode ser dividido por outros números. Usamos o % para ver se a divisão tem resto 0. Se tiver, significa que ele é divisível por outro número e então não é primo. O break serve para parar o for assim que encontramos um divisor. Se o número for menor que 2, ele também não é primo.

```python
numero = int(input("Digite um número: "))

if numero < 2:
    primo = False
else:
    primo = True
    for i in range(2, int(numero ** 0.5) + 1):
        if numero % i == 0:
            primo = False
            break   # achou um divisor, não precisa continuar

if primo:
    print(numero, "é primo")
else:
    print(numero, "não é primo")
```

19. Jogo de Adivinhação

Implemente um jogo onde o usuário deve adivinhar um número
gerado aleatoriamente pelo computador.

Conceitos que caem aqui: random.randint(), while True, break e contador de tentativas

Minha explicação: Primeiro usamos a biblioteca random para sortear um número de 1 a 100. Depois usamos o while True para deixar o jogo repetindo até a pessoa acertar. A cada tentativa, o usuário digita um número e o código verifica se ele é maior ou menor que o número sorteado, dando uma dica. Também usamos um contador para saber quantas tentativas a pessoa fez. 20. Matriz Transposta

Dada uma matriz (lista de listas), crie sua matriz transposta.

Conceitos que caem aqui: listas dentro de listas, dois for encaixados, len(matriz) e len(matriz[0])

Minha explicação: Primeiro temos uma matriz, que é basicamente uma lista que possui outras listas dentro dela, formando linhas e colunas. Transpor a matriz significa trocar as linhas pelas colunas. Nesse código, usamos dois for: um fica responsável por passar pelas colunas e o outro pelas linhas. O len(matriz) serve para saber quantas linhas temos e o len(matriz[0]) para saber quantas colunas tem. No final, juntamos os valores em uma nova matriz com as linhas e colunas trocadas.

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

transposta = []
for coluna in range(len(matriz[0])):
    nova_linha = []
    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])
    transposta.append(nova_linha)

print("Matriz original:")
for linha in matriz:
    print(linha)

print("Matriz transposta:")
for linha in transposta:
    print(linha)
```

Conceitos que caem aqui: lista ordenada, índices, // e -1

Minha explicação: Primeiro precisamos ter uma lista ordenada, porque a busca binária depende disso. A ideia é pegar o número que está no meio da lista e comparar com o número que estamos procurando. Se for igual, encontramos. Se o número que procuramos for menor, ignoramos a metade da direita. Se for maior, ignoramos a metade da esquerda. Assim, vamos diminuindo a lista que precisamos procurar, ficando mais rápido do que olhar número por número. Se encontrar, o código retorna a posição dele na lista. Se não encontrar, retorna -1.

```python
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1   # não encontrou


numeros = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(busca_binaria(numeros, 23))
print(busca_binaria(numeros, 7))
```

22. Criptografia de Cifra de César

Crie uma função para criptografar uma mensagem usando a cifra de César.

Conceitos que caem aqui: ord(), chr() e % 26

Minha explicação: A Cifra de César funciona trocando cada letra por outra que está algumas posições à frente no alfabeto. Primeiro usamos o ord() para pegar um número que representa a letra. Depois fazemos a conta com o número de posições que queremos andar e usamos o chr() para transformar esse número de volta em letra. O % 26 serve para voltar para o começo do alfabeto quando passamos do Z. Espaços, números e pontuação ficam do jeito que estão.

```python
def criptografar(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        if letra.isupper():
            resultado += chr((ord(letra) - 65 + deslocamento) % 26 + 65)
        elif letra.islower():
            resultado += chr((ord(letra) - 97 + deslocamento) % 26 + 97)
        else:
            resultado += letra   # espaço, pontuação etc continuam iguais

    return resultado


texto = input("Digite a mensagem: ")
chave = int(input("Digite o deslocamento (ex: 3): "))

print("Mensagem criptografada:", criptografar(texto, chave))
```

Implemente uma função para validar um CPF brasileiro.

Conceitos que caem aqui: replace(), for, %, if/else e validação

Minha explicação: Primeiro pedimos o CPF e usamos o replace() para tirar os pontos e o traço, deixando só os números. Depois verificamos se ele tem 11 números. Os dois últimos números são os dígitos verificadores, que servem para conferir se o CPF está correto.

Primeiro pegamos os 9 primeiros números, fazemos as multiplicações com os pesos de 10 até 2 e usamos uma conta para descobrir o primeiro dígito. Depois fazemos praticamente a mesma coisa com os 10 primeiros números, usando os pesos de 11 até 2, para descobrir o segundo dígito. No final, comparamos os dois dígitos calculados com os dois últimos números do CPF. Se forem iguais, o CPF é considerado válido.

Também verificamos se todos os números são iguais, porque um CPF como 111.111.111-11 pode passar pela conta, mas não é um CPF válido.

```python
def validar_cpf(cpf):
    # tira ponto e traço, deixando só os números
    cpf = cpf.replace(".", "").replace("-", "").strip()

    if len(cpf) != 11 or not cpf.isdigit():
        return False

    if cpf == cpf[0] * 11:   # casos tipo 00000000000
        return False

    # primeiro dígito verificador
    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)
    digito1 = (soma * 10) % 11
    if digito1 == 10:
        digito1 = 0

    # segundo dígito verificador
    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)
    digito2 = (soma * 10) % 11
    if digito2 == 10:
        digito2 = 0

    return digito1 == int(cpf[9]) and digito2 == int(cpf[10])


documento = input("Digite um CPF: ")

if validar_cpf(documento):
    print("CPF válido")
else:
    print("CPF inválido")
```

```python
import random

palavras = ["garrafa", "gigi", "teclado", "celular", "faculdade"]
palavra = random.choice(palavras)

letras_certas = []
tentativas = []
vidas = 6

while vidas > 0:
    # monta como a palavra aparece na tela
    exibicao = ""
    for letra in palavra:
        if letra in letras_certas:
            exibicao += letra + " "
        else:
            exibicao += "_ "

    print(exibicao)
    print("Vidas:", vidas)

    if "_" not in exibicao:
        print("Parabéns, você acertou a palavra:", palavra)
        break

    chute = input("Digite uma letra: ").lower()

    if chute in tentativas:
        print("Você já tentou essa letra")
        continue

    tentativas.append(chute)

    if chute in palavra:
        letras_certas.append(chute)
        print("Boa, tem essa letra!")
    else:
        vidas -= 1
        print("Essa letra não tem")

if vidas == 0:
    print("Suas vidas acabaram. A palavra era:", palavra)
```

31. Ordenação por Bolha

Implemente o algoritmo de ordenação por bolha (Bubble Sort).

Conceitos que caem aqui: dois for, comparação e troca de posições

Minha explicação: Primeiro temos uma lista de números que queremos colocar em ordem. Depois o primeiro for faz as rodadas da ordenação. Dentro dele, o segundo for vai comparando os números que estão um ao lado do outro. Se o número da esquerda for maior que o da direita, trocamos eles de lugar. Quando termina uma rodada, o maior número já fica no final da lista. Por isso, na próxima rodada, não precisamos comparar esse número novamente. O processo continua até todos os números ficarem na ordem certa.

```python
def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista


numeros = [64, 34, 25, 12, 22, 11, 90]
print("Antes:", numeros)
print("Depois:", bubble_sort(numeros.copy()))
```

Depois de cada jogada, usamos um if para trocar o jogador, passando de X para O e depois de O para X. Também verificamos se o tabuleiro ficou cheio sem ninguém ganhar, para identificar o empate.

```python
def mostrar(tabuleiro):
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("---------")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("---------")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])


def venceu(tabuleiro, jogador):
    combinacoes = [[0, 1, 2], [3, 4, 5], [6, 7, 8],
                   [0, 3, 6], [1, 4, 7], [2, 5, 8],
                   [0, 4, 8], [2, 4, 6]]
    for c in combinacoes:
        if tabuleiro[c[0]] == jogador and tabuleiro[c[1]] == jogador and tabuleiro[c[2]] == jogador:
            return True
    return False


tabuleiro = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
jogador = "X"
jogadas = 0

while True:
    mostrar(tabuleiro)
    posicao = input("Jogador " + jogador + ", escolha uma posição (1 a 9): ")

    if posicao not in ["1", "2", "3", "4", "5", "6", "7", "8", "9"]:
        print("Posição inválida")
        continue

    indice = int(posicao) - 1

    if tabuleiro[indice] == "X" or tabuleiro[indice] == "O":
        print("Essa posição já foi usada")
        continue

    tabuleiro[indice] = jogador
    jogadas += 1

    if venceu(tabuleiro, jogador):
        mostrar(tabuleiro)
        print("O jogador", jogador, "venceu!")
        break

    if jogadas == 9:
        mostrar(tabuleiro)
        print("Deu velha! Ninguém venceu")
        break

    # troca de jogador
    if jogador == "X":
        jogador = "O"
    else:
        jogador = "X"
```

No final, essa lista vai ter todos os números primos até o número escolhido.

45. Par ou Ímpar em Lista

Classifique os números de uma lista como pares ou ímpares.

Conceitos que caem aqui: for, % e if/else

Minha explicação: Primeiro temos uma lista com vários números. Depois usamos o for para passar por cada número da lista, um por um. Para descobrir se ele é par ou ímpar, usamos o % para ver o resto da divisão por 2. Se o resto for 0, colocamos o número na lista de pares. Se não for, colocamos na lista de ímpares.

```python
numeros = [11, 4, 7, 18, 25, 30, 3]

pares = []
impares = []

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print("Pares:", pares)
print("Ímpares:", impares)
```

46. Calcular Média e Desvio Padrão

Calcule a média e o desvio padrão de uma lista de números.

Conceitos que caem aqui: média, variância, raiz quadrada e statistics

Minha explicação: Primeiro calculamos a média dos números da lista. Depois vemos quanto cada número está distante dessa média. Elevamos essas diferenças ao quadrado para não ficarem negativas e fazemos a média desses resultados, que é a variância. Por último, tiramos a raiz quadrada da variância para chegar no desvio padrão. O desvio padrão mostra o quanto os números estão espalhados em relação à média. No final, usamos o statistics só para conferir se o resultado está certo.

```python
import statistics

numeros = [10, 12, 23, 23, 16, 23, 21, 16]

media = sum(numeros) / len(numeros)

soma_quadrados = 0
for n in numeros:
    soma_quadrados += (n - media) ** 2

variancia = soma_quadrados / len(numeros)
desvio = variancia ** 0.5

print("Média:", round(media, 2))
print("Desvio padrão:", round(desvio, 2))

# conferindo com a biblioteca pronta
print("Conferindo com statistics:", round(statistics.pstdev(numeros), 2))
```

47. Contar Números em Lista

Conte quantas vezes cada número aparece em uma lista.

Conceitos que caem aqui: dicionário e .get()

Minha explicação: Primeiro temos uma lista com vários números. Depois usamos um dicionário para guardar cada número e a quantidade de vezes que ele aparece. No final, temos cada número junto com a quantidade de vezes que apareceu.

```python
numeros = [1, 2, 2, 3, 3, 3, 4, 1, 5, 3]

contagem = {}
for n in numeros:
    contagem[n] = contagem.get(n, 0) + 1

for numero, vezes in contagem.items():
    print("O número", numero, "aparece", vezes, "vez(es)")
```

48. Ordenação por Inserção

Implemente o algoritmo de ordenação por inserção (Insertion Sort).

Conceitos que caem aqui: for, while, comparação e troca de posições

Minha explicação: Primeiro temos uma lista de números que queremos organizar. O for vai pegando um número por vez e comparando com os números que já estão organizados. Se o número da esquerda for maior, usamos o while para ir empurrando ele para trás até encontrar o lugar certo. É parecido com quando a gente organiza cartas na mão: pega uma carta e coloca no lugar certo entre as que já estão organizadas. No final, a lista fica toda em ordem.

```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        atual = lista[i]
        j = i - 1

        # empurra pra direita todo mundo que for maior que o atual
        while j >= 0 and lista[j] > atual:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = atual
    return lista


numeros = [8, 3, 15, 1, 9, 4]
print("Antes:", numeros)
print("Depois:", insertion_sort(numeros.copy()))
```

49. Soma dos Dígitos

Calcule a soma dos dígitos de um número inteiro fornecido.

Conceitos que caem aqui: str(), int(), abs() e for

Minha explicação: Primeiro pegamos o número que o usuário digitou e usamos o abs() para deixar ele positivo, caso tenha sido digitado um número negativo. Depois usamos o str() para transformar o número em texto, assim conseguimos passar por cada dígito separadamente. Dentro do for, transformamos cada dígito novamente em número usando o int() e vamos somando todos eles. No final, temos a soma dos dígitos do número.

```python
numero = int(input("Digite um número inteiro com dois digitos: "))

soma = 0
for digito in str(abs(numero)):
    soma += int(digito)

print("A soma dos dígitos é:", soma)
```

50. Cálculo de Frequência de Letras

Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

Conceitos que caem aqui: .lower(), .isalpha(), dicionário, .get() e sorted()

Minha explicação: Primeiro pedimos uma frase e usamos o .lower() para deixar todas as letras minúsculas. Depois o for passa por cada caractere da frase. Usamos o .isalpha() para verificar se o caractere é uma letra, assim ele ignora espaços, números e pontuação. Depois usamos um dicionário para guardar cada letra e quantas vezes ela apareceu. O .get() ajuda a pegar a quantidade que a letra já tem e somar 1. No final, usamos o sorted() para mostrar as letras em ordem.

```python
texto = input("Digite uma frase: ").lower()

frequencia = {}
for caractere in texto:
    if caractere.isalpha():   # só conta letra, ignora espaço e pontuação
        frequencia[caractere] = frequencia.get(caractere, 0) + 1

for letra in sorted(frequencia):
    print(letra, "->", frequencia[letra])
```
