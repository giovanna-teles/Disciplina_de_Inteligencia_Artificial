###19.Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

1. Gero dois arrays de 20 números aleatórios cada, pensando neles como vetores no espaço.
2. Calculo o produto escalar entre eles com `np.dot(array1, array2)` .. essa é uma conta que mistura os dois vetores num único número.
3. Calculo também o 'tamanho' (norma) de cada vetor com `np.linalg.norm`.
4. A fórmula da trigonometria diz que o cosseno do ângulo entre dois vetores é o produto escalar dividido pelo produto dos tamanhos deles. Uso o `np.clip` só de segurança, pra garantir que o valor não escape um pouquinho de -1 a 1 por erro de arredondamento (o que quebraria a próxima conta).
5. Por fim, uso `np.arccos` pra transformar o cosseno de volta num ângulo (em radianos) e `np.degrees` pra converter esse ângulo pra graus, que é mais fácil de entender.

```python
import numpy as np

np.random.seed(42)
array1 = np.random.randint(1, 50, size=20)
array2 = np.random.randint(1, 50, size=20)

produto_escalar = np.dot(array1, array2)
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)

cos_theta = produto_escalar / (norma1 * norma2)
angulo_rad = np.arccos(np.clip(cos_theta, -1.0, 1.0))
angulo_graus = np.degrees(angulo_rad)

print("Cosseno do ângulo:", cos_theta)
print("Ângulo entre os vetores (graus):", angulo_graus)
```

### 2. Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.

1. Uma equação diferencial é aquela que descreve 'como uma coisa muda', em vez de dizer direto o valor dela. Pra resolver isso, uso a função `solve_ivp` (que significa 'resolver problema de valor inicial' haha).
2. Escrevo a equação como uma função Python: `dy/dx = x - y` vira `def eq_diferencial(t, y): return t - y` (troquei x por t só porque é a letra que o SciPy espera pro 'tempo' ou variável independente).
3. Chamo `solve_ivp` passando: a função da equação, o intervalo `[0, 10]` e a condição inicial `[1]` (que é o valor de y quando x=0).
4. Uso `t_eval=np.linspace(0, 10, 100)` pra pedir 100 pontos igualmente espaçados dentro desse intervalo, só pra ter uma curva bonitinha e detalhada como resposta.
5. No final, `solucao.t` são os pontos de tempo e `solucao.y[0]` é a solução de y em cada um desses pontos.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(t, y):
    return t - y  # dy/dx = x - y (aqui x é representado por t)

solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print("Primeiros valores de t:", solucao.t[:5])
print("Primeiros valores de y:", solucao.y[0][:5])
print("Valor final de y em t=10:", solucao.y[0][-1])
```

### 3. Encontre as raízes da função f(x) = x⁴ - 3x³ + 2 utilizando SciPy.


1. 'Raiz de uma função' é o valor de x que faz ela dar zero. Como essa função pode ter mais de uma raiz, uso a `fsolve`, que busca uma raiz 'perto' de um chute inicial que eu dou.
2. Defino a função com a fórmula do exercício.
3. Como uma função de grau 4 pode ter até 4 raízes diferentes, eu chuto vários pontos de partida diferentes (`-2, 0, 1, 3`) e rodo a `fsolve` pra cada um — é tipo jogar várias 'iscas' em pontos diferentes da reta pra ver onde cada uma 'pesca' uma raiz.
4. Depois eu arredondo os resultados e uso `set()` (conjunto) pra tirar raízes repetidas, caso dois chutes diferentes caiam na mesma raiz.

```python
from scipy.optimize import fsolve
import numpy as np

def f(x):
    return x**4 - 3*x**3 + 2

chutes_iniciais = [-2, 0, 1, 3]
raizes = [fsolve(f, x0)[0] for x0 in chutes_iniciais]
raizes_unicas = sorted(set(np.round(raizes, 4)))

print("Raízes encontradas:", raizes_unicas)
```

### 4.Use a função de otimização de SciPy para maximizar f(x) = -x² + 4x + 1.

1. O SciPy só tem uma função pronta pra achar o **mínimo** de uma função (`minimize`), não pra achar o máximo direto.
2. O truque clássico aqui é matemático e simples: maximizar f(x) é a mesma coisa que minimizar ..f(x) (o ponto mais alto de uma curva vira o ponto mais baixo se eu virar ela de cabeça pra baixo). Por isso crio uma segunda função, `f_negativa`, que é só `f(x)` com o sinal trocado.
3. Chamo `minimize(f_negativa, x0=0)`, começando a busca em x=0.
4. No final, o `x` que a `minimize` encontrou é o mesmo ponto de máximo da função original — só preciso usar `f(x)` (a função original, sem o sinal trocado) pra saber o valor máximo de verdade.

```python
from scipy.optimize import minimize

def f(x):
    return -x**2 + 4*x + 1

# Para maximizar f(x), minimizamos -f(x)
def f_negativa(x):
    return -f(x)

resultado = minimize(f_negativa, x0=0)
print("Ponto de máximo em x =", resultado.x[0])
print("Valor máximo de f(x):", f(resultado.x[0]))
```

2. Defino a função com os dois parâmetros .. e aqui teve um detalhe que achei estranho no começo: o SciPy espera a função escrita como `f(y, x)`, com o y vindo primeiro, não `f(x, y)`. É só uma convenção da própria biblioteca.
3. Uso `integrate.dblquad(f, 0, 2, lambda x: 0, lambda x: 1)`: os dois primeiros números (`0, 2`) são os limites de x, e os dois `lambda x: ...` são os limites de y (aqui escritos como funções 'fixas' porque não dependem de x, mas a função pede esse formato).

###16.Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0) = 1 e y'(0) = 0.

1. O `solve_ivp` só sabe resolver equações 'de primeira ordem' (que só têm uma derivada). Como essa equação tem uma segunda derivada (d²y/dx²), o truque é quebrar ela em duas equações mais simples: uma pra `y` e outra pra `dy` (a velocidade de variação de y).
2. Dentro da função, `y, dy = estado` separa essas duas partes, e eu calculo `d2y = -y` (que é exatamente o que a equação original pedia) e devolvo as duas taxas de variação juntas: `[dy, d2y]`.
3. As condições iniciais também viram duas: `[1, 0]`, que representam y(0)=1 e a 'velocidade inicial' y'(0)=0.
4. Fiz um comentário no código lembrando que a solução exata dessa equação é `y(t) = cos(t)` .. dá pra comparar os números que saíram com o cosseno pra conferir se bateu.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_segunda_ordem(t, estado):
    y, dy = estado
    d2y = -y
    return [dy, d2y]

solucao = solve_ivp(eq_segunda_ordem, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100))

print("Primeiros valores de t:", solucao.t[:5])
print("Solução y(t):", solucao.y[0][:5])
# Solução analítica esperada: y(t) = cos(t)
```

### 18.Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

1. Defino uma lista de valores que mistura positivos e negativos de propósito.
2. Uso um truque de uma linha só (chamado 'list comprehension') pra montar a lista de cores automaticamente: `'green' if v >= 0 else 'red' for v in valores`..em português seria 'pra cada valor v na lista, se ele for maior ou igual a zero use verde, senão use vermelho'.
3. Desenho as barras normalmente passando essa lista de cores no `color=cores`.
4. Adiciono `plt.axhline(0, color='black', ...)` pra desenhar uma linha horizontal exatamente no zero, o que ajuda a visualizar melhor quais barras estão 'acima' e quais estão 'abaixo' da linha de base.

```python
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D', 'E']
valores = [10, -5, 8, -3, 6]
cores = ['green' if v >= 0 else 'red' for v in valores]

plt.bar(categorias, valores, color=cores)
plt.title("Gráfico de Barras com Valores Positivos e Negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.axhline(0, color='black', linewidth=0.8)
plt.show()
```

### 20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x².


1. Importo uma ferramenta extra, `MultipleLocator`, de dentro de `matplotlib.ticker` .. ela serve pra controlar de quanto em quanto as marcações (e as linhas da grade) aparecem no eixo.
2. Dessa vez, em vez de usar `plt.plot()` direto, crio explicitamente uma `fig` (a figura toda) e um `ax` (o eixo dentro dela) com `plt.subplots()` ..isso me dá mais controle fino sobre os detalhes, que é exatamente o que preciso pra personalizar a grade.
3. Desenho a curva `y = x²` nesse eixo com `ax.plot(x, y)`.
4. `ax.xaxis.set_major_locator(MultipleLocator(2))` diz 'coloca uma marcação a cada 2 unidades no eixo x', e faço o mesmo no eixo y mas de 20 em 20, já que os valores de y ali crescem bem mais rápido (é x ao quadrado, então até x=10 o y já chega a 100).
5. Por fim, `ax.grid(True, linestyle='--')` liga a grade usando exatamente esse espaçamento personalizado que acabei de definir.

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
import numpy as np

x = np.linspace(-10, 10, 200)
y = x**2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Gráfico de Linha com Grade Personalizada - y = x²")
ax.set_xlabel("x")
ax.set_ylabel("y")

ax.xaxis.set_major_locator(MultipleLocator(2))
ax.yaxis.set_major_locator(MultipleLocator(20))
ax.grid(True, linestyle='--')

plt.show()
```
